```python
#!/usr/bin/env python
# coding: utf-8
import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_parquet("https://d37ci6vzurychx.cloudfront.net/trip-data/green_tripdata_2025-11.parquet")
```

```python
engine = create_engine('postgresql://postgres:postgres@localhost:5433/ny_taxi')
```

```python
#print(pd.io.sql.get_schema(df, name='green_trip_data', con=engine))
```

```python
df.head(0).to_sql(name='green_trip_data',con=engine, if_exists='replace')
```

```python
df.to_sql(name='green_trip_data',con=engine, if_exists='append')
```

```python
df_zoon = pd.read_csv("https://github.com/DataTalksClub/nyc-tlc-data/releases/download/misc/taxi_zone_lookup.csv")
```

```python
df_zoon.head(0).to_sql(name='taxi_zoon_lookup',con=engine, if_exists='replace')
```

```python
df_zoon.to_sql(name='taxi_zoon_lookup',con=engine, if_exists='append')
```
